```python
import matplotlib.pyplot as plt

# File sizes in MB
file_sizes = [200, 400, 600, 800, 1000]

# Time taken in seconds for each language
time_c = [18, 26, 39, 50, 59]
time_cpp = [21, 33, 42, 59, 65]
time_java = [24, 37, 47, 65, 73]
time_r = [29, 40, 51, 69, 84]
time_python = [35, 45, 59, 75, 100]

# Plotting
plt.figure(figsize=(10, 6))
plt.plot(file_sizes, time_c, marker='o', label='C')
plt.plot(file_sizes, time_cpp, marker='s', label='C++')
plt.plot(file_sizes, time_java, marker='^', label='Java')
plt.plot(file_sizes, time_r, marker='D', label='R')
plt.plot(file_sizes, time_python, marker='x', label='Python')

plt.title("Performance Comparison (Uppercase Conversion)")
plt.xlabel("File Size (MB)")
plt.ylabel("Time Taken (Seconds)")
plt.legend()
plt.grid(True)
plt.xticks(file_sizes)
plt.tight_layout()

# Save and show
plt.savefig("performance_comparison.png")
plt.show()
```
